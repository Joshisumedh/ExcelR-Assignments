7.	Given a number, check if it's positive, and if it is also even. If not positive, print if it's zero or negative

```python
num = int(input("Enter a number: "))

if num > 0:
    if num % 2 == 0:
        print("The number is positive and even.")
    else:
        print("The number is positive but odd.")
elif num == 0:
    print("The number is zero.")
else:
    print("The number is negative.")
```

11.	Create a small DataFrame manually with 10 rows and columns: 'Name', 'Age', 'City', and 'Salary'. Then:
- Use .info(), .describe()
- Select 'Name' and 'City' columns
- Drop 'City' column
- Fill any missing values in 'Salary' column with the mean
- Remove any duplicate rows


```python
data = {
    'Name': ['Sumedh', 'pranav', 'Ananya', 'Sukhada', 'vedant', 'pavan', 'Atharva', 'Vikram', 'varad', 'yash'],
    'Age': [25, 30, 22, 28, 25, 35, 27, 40, 32, 22],
    'City': ['Pune', 'Mumbai', 'Delhi', 'Pune', 'Pune', 'Bangalore', 'Mumbai', 'Delhi', 'Chennai', 'Delhi'],
    'Salary': [50000, 60000, np.nan, 55000, 50000, 65000, np.nan, 70000, 62000, 55000]
}
```

```python
df = pd.DataFrame(data)
```

```python
#.info
print("DataFrame Info:")
print(df.info(),)
```

```python
#.describe
print("DataFrame Description:")
print(df.describe())
```

```python
#Select 'Name' and 'City' columns
print("🔹 Selected Columns (Name & City):")
print(df[['Name', 'City']])
```

```python
#Drop 'City' column
df = df.drop(columns=['City'])
print("After Dropping 'City' Column:")
print(df)
```

```python
#Fill any missing values in 'Salary' column with the mean
df['Salary'].fillna(df['Salary'].mean(), inplace=True)
print("After Filling Missing Salary Values:")
print(df)
```

```python
#Remove any duplicate rows
df = df.drop_duplicates()
print("After Removing Duplicates:")
print(df)
```
