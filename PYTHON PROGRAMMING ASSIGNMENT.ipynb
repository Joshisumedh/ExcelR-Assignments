Exercise 1: Prime NumbersWrite a Python program that checks whether a given number is prime or not. A prime number is a natural number greater than 1 that has no positive divisors other than 1 and itself.

```python
# Check whether a number is prime or not

num = int(input("Enter a number: "))

if num > 1:
    
    for i in range(2, num):
        if num % i == 0:
            print(num, "is not a prime number.")
            break
    else:
        print(num, "is a prime number.")
else:
    print(num, "is not a prime number.")
```

 Exercise 2: Product of Random Numbers
Develop a Python program that generates two random numbers and asks the user to enter the product of these numbers. The program should then check if the user's answer is correct and display an appropriate message.


```python
import random

# Generate two random integers (any number in Python's integer range)
num1 = random.randint(-100, 100)
num2 = random.randint(-100, 100)


print(f"What is {num1} × {num2}?")
user_answer = int(input("Enter your answer: "))


correct_answer = num1 * num2


if user_answer == correct_answer:
    print("✅ Correct! Well done!")
else:
    print(f"❌ Incorrect. The correct answer is {correct_answer}.")
```

Exercise 3: Squares of Even/Odd Numbers
Create a Python script that prints the squares of all even or odd numbers within the range of 100 to 200. Choose either even or odd numbers and document your choice in the code.


```python
choice = input("Do you want squares of 'even' or 'odd' numbers? ").strip().lower()

# Loop through numbers from 100 to 200 (inclusive)
for number in range(100, 201):
    if choice == "even" and number % 2 == 0:
        print(f"The square of {number} is {number ** 2}")
    elif choice == "odd" and number % 2 != 0:
        print(f"The square of {number} is {number ** 2}")


if choice not in ["even", "odd"]:
    print("Invalid choice! Please enter 'even' or 'odd'.")
```

Exercise 4:
✅ Problem Statement (Question): Hospital Billing System
A hospital charges its patients based on the following:
•	Room charges per day: ₹2,000
•	Doctor consultation fee (flat): ₹1,500
•	Lab test charges: Based on the number of tests taken. Each test costs ₹300
•	Medicine charges: Total cost of medicines provided
•	Discount: If the total bill (before discount) exceeds ₹10,000, a 10% discount is applied
Write a Python program to calculate the final bill for a patient who:
•	Stayed for 4 days
•	Had 3 lab tests
•	Had medicine charges of ₹2,400


```python
# Exercise 4: Hospital Billing System

# Given data
room_charge_per_day = 2000
doctor_fee = 1500
lab_test_charge_per_test = 300
discount_rate = 0.10  # 10%

# Patient details
days_stayed = 4
num_lab_tests = 3
medicine_charges = 2400


room_charges = room_charge_per_day * days_stayed
lab_test_charges = lab_test_charge_per_test * num_lab_tests


total_before_discount = room_charges + doctor_fee + lab_test_charges + medicine_charges


if total_before_discount > 10000:
    discount = total_before_discount * discount_rate
else:
    discount = 0


final_bill = total_before_discount - discount

# Step 5: Display the bill details
print("----- Hospital Billing System -----")
print(f"Room Charges (₹{room_charge_per_day}/day × {days_stayed} days): ₹{room_charges}")
print(f"Doctor Consultation Fee: ₹{doctor_fee}")
print(f"Lab Test Charges (₹{lab_test_charge_per_test}/test × {num_lab_tests} tests): ₹{lab_test_charges}")
print(f"Medicine Charges: ₹{medicine_charges}")
print(f"Total Before Discount: ₹{total_before_discount}")
print(f"Discount Applied: ₹{discount}")
print("-----------------------------------")
print(f"Final Bill Amount: ₹{final_bill}")
print("-----------------------------------")
```
